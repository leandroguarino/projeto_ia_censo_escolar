```python
import pandas as pd
import os
```

```python
colunas = [
    'NU_ANO_CENSO',
    'CO_ENTIDADE',
    'QT_MAT_BAS', 'QT_MAT_INF', 'QT_MAT_INF_CRE', 'QT_MAT_INF_PRE',
    'QT_MAT_FUND', 'QT_MAT_FUND_AI', 'QT_MAT_FUND_AF',
    'QT_MAT_MED', 'QT_MAT_PROF', 'QT_MAT_PROF_TEC',
    'QT_MAT_EJA',
    'QT_DOC_INF', 'QT_DOC_INF_CRE', 'QT_DOC_INF_PRE',
    'QT_DOC_FUND', 'QT_DOC_FUND_AI', 'QT_DOC_FUND_AF',
    'QT_DOC_MED', 'QT_DOC_PROF', 'QT_DOC_PROF_TEC',
    'QT_DOC_EJA',
    'IN_EJA'
]
```

```python
caminhos = {
    2022: '../data/microdados_censo_escolar_2022/microdados/dados/microdados_ed_basica_2022.csv',
    2023: '../data/microdados_censo_escolar_2023/dados/microdados_ed_basica_2023.csv',
    2024: '../data/microdados_censo_escolar_2024/dados/microdados_ed_basica_2024.csv'
}
```

```python
dfs = []
for ano, caminho in caminhos.items():
    print(f'Lendo dados de {ano}...')
    df_ano = pd.read_csv(caminho, encoding='latin-1', sep=';', low_memory=False, usecols=colunas)
    dfs.append(df_ano)
    print(f'{ano}: {df_ano.shape[0]} escolas carregadas')
```

```python
df = pd.concat(dfs, ignore_index=True)
print(f'DataFrame consolidado: {df.shape[0]} registros x {df.shape[1]} colunas')
df.head()
```

```python
# Filtrar escolas com EJA
df_eja = df[df['IN_EJA'] == 1].copy()
print(f'Total de escolas com EJA: {len(df_eja)}')
df_eja.head()
```

```python
# Contagem de escolas com EJA por ano
df_eja['NU_ANO_CENSO'].value_counts().sort_index()
```

```python
# Estatísticas descritivas das escolas com EJA
df_eja.describe()
```

```python
# DataFrame sem as colunas NU_ANO_CENSO, CO_ENTIDADE e IN_EJA
df_features = df_eja.drop(columns=['NU_ANO_CENSO', 'CO_ENTIDADE', 'IN_EJA', 'QT_MAT_BAS'])
print(f'Colunas removidas. DataFrame atual: {df_features.shape[0]} registros x {df_features.shape[1]} colunas')
df_features.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
import numpy as np

# Preparar dados: remover linhas com NaN na coluna alvo e preencher demais NaNs com 0
df_model = df_features.dropna(subset=['QT_MAT_EJA']).fillna(0)

# Separar features (X) e target (y)
X = df_model.drop(columns=['QT_MAT_EJA'])
y = df_model['QT_MAT_EJA']

# Dividir em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Dicionário de regressores
regressores = {
    'LinearRegression': LinearRegression(),
    'RandomForestRegressor': RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    'GradientBoostingRegressor': GradientBoostingRegressor(random_state=42),
    'ExtraTreesRegressor': ExtraTreesRegressor(n_estimators=100, random_state=42, n_jobs=-1)
}

# Treinar e avaliar cada modelo
for nome, modelo in regressores.items():
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    
    print(f'\n{nome}:')
    print(f'  MAE : {mae:,.2f}')
    print(f'  RMSE: {rmse:,.2f}')
    print(f'  R²  : {r2:.4f}')
```

```python
# Importância das features do RandomForestRegressor
rf = regressores['RandomForestRegressor']
importancias = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)
print('Top 10 features mais importantes:')
print(importancias.head(10))
```

```python
import matplotlib.pyplot as plt

# Gráfico das 10 features mais importantes
plt.figure(figsize=(10, 6))
importancias.head(10).plot(kind='barh', color='#2E86AB')
plt.title('Top 10 Features Mais Importantes para Prever QT_MAT_EJA')
plt.xlabel('Importância')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Comparar valores reais vs preditos (RandomForest)
y_pred_rf = regressores['RandomForestRegressor'].predict(X_test)

plt.figure(figsize=(8, 8))
plt.scatter(y_test, y_pred_rf, alpha=0.3, s=10)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Valores Reais (QT_MAT_EJA)')
plt.ylabel('Valores Preditos (QT_MAT_EJA)')
plt.title('Valores Reais vs Preditos - RandomForestRegressor')
plt.tight_layout()
plt.show()
```

```python
import pickle

# Salvar o melhor modelo (RandomForest) e os dados de teste
with open('../data/modelo_rf_eja.pkl', 'wb') as f:
    pickle.dump(regressores['RandomForestRegressor'], f)

print('Modelo salvo em ../data/modelo_rf_eja.pkl')
```

```python
# Análise de resíduos
residuos = y_test - y_pred_rf

fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Histograma dos resíduos
axs[0].hist(residuos, bins=50, color='#2E86AB', edgecolor='black')
axs[0].set_title('Distribuição dos Resíduos')
axs[0].set_xlabel('Resíduo')
axs[0].set_ylabel('Frequência')

# Resíduos vs Valores Preditos
axs[1].scatter(y_pred_rf, residuos, alpha=0.3, s=10)
axs[1].axhline(0, color='r', linestyle='--')
axs[1].set_title('Resíduos vs Valores Preditos')
axs[1].set_xlabel('Valores Preditos')
axs[1].set_ylabel('Resíduo')

plt.tight_layout()
plt.show()
```
