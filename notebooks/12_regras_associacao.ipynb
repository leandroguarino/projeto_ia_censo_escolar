```python
import pandas as pd
import os
```

```python
colunas = [
    'NU_ANO_CENSO',
    'CO_ENTIDADE',
    'QT_MAT_BAS', 'QT_MAT_INF', 'QT_MAT_INF_CRE', 'QT_MAT_INF_PRE',
    'QT_MAT_FUND', 'QT_MAT_FUND_AI', 'QT_MAT_FUND_AF',
    'QT_MAT_MED', 'QT_MAT_PROF', 'QT_MAT_PROF_TEC',
    'QT_MAT_EJA',
    'QT_DOC_INF', 'QT_DOC_INF_CRE', 'QT_DOC_INF_PRE',
    'QT_DOC_FUND', 'QT_DOC_FUND_AI', 'QT_DOC_FUND_AF',
    'QT_DOC_MED', 'QT_DOC_PROF', 'QT_DOC_PROF_TEC',
    'QT_DOC_EJA',
    'IN_EJA'
]
```

```python
caminhos = {
    2022: '../data/microdados_censo_escolar_2022/microdados/dados/microdados_ed_basica_2022.csv',
    2023: '../data/microdados_censo_escolar_2023/dados/microdados_ed_basica_2023.csv',
    2024: '../data/microdados_censo_escolar_2024/dados/microdados_ed_basica_2024.csv'
}
```

```python
dfs = []
for ano, caminho in caminhos.items():
    print(f'Lendo dados de {ano}...')
    df_ano = pd.read_csv(caminho, encoding='latin-1', sep=';', low_memory=False, usecols=colunas)
    dfs.append(df_ano)
    print(f'{ano}: {df_ano.shape[0]} escolas carregadas')
```

```python
df = pd.concat(dfs, ignore_index=True)
print(f'DataFrame consolidado: {df.shape[0]} registros x {df.shape[1]} colunas')
df.head()
```

```python
# Filtrar escolas com EJA
df_eja = df[df['IN_EJA'] == 1].copy()
print(f'Total de escolas com EJA: {len(df_eja)}')
df_eja.head()
```

```python
# Eliminar colunas CO_ENTIDADE e IN_EJA
df_eja = df_eja.drop(columns=['CO_ENTIDADE', 'IN_EJA'])
print(f'Colunas eliminadas: CO_ENTIDADE, IN_EJA')
print(f'DataFrame após remoção: {df_eja.shape[0]} registros x {df_eja.shape[1]} colunas')
print(f'Colunas restantes: {list(df_eja.columns)}')
```

```python
# Criar colunas categóricas baseadas em faixas (discretização por quantis)
# Método estatístico: Quantile-based discretization
# Isso divide os dados em bins baseados em quantis, garantindo distribuição uniforme

import numpy as np

colunas_numericas = [col for col in df_eja.columns if col != 'NU_ANO_CENSO']
num_bins = 4  # Número de faixas (quartis)

print(f'Criando faixas categóricas para {len(colunas_numericas)} colunas numéricas usando {num_bins} quantis...\n')

for col in colunas_numericas:
    # Criar nome da nova coluna categórica
    col_cat = f'{col}_FAIXA'
    
    # Converter para numeric (caso esteja como string)
    col_numeric = pd.to_numeric(df_eja[col], errors='coerce')
    
    # Pular colunas com todos valores NaN
    if col_numeric.isna().all():
        print(f'  {col} -> pulada (todos valores NaN)')
        continue
    
    # Calcular os quantis manualmente
    quantiles = [i / num_bins for i in range(num_bins + 1)]
    bin_edges = col_numeric.quantile(quantiles).values
    
    # Remover NaNs e duplicatas, garantindo monotonicidade
    bin_edges = bin_edges[~np.isnan(bin_edges)]
    bin_edges = np.unique(bin_edges)
    
    # Se tiver menos de 2 bins únicos, usar método alternativo
    if len(bin_edges) < 2:
        # Usar range min-max com divisões iguais
        min_val = col_numeric.min()
        max_val = col_numeric.max()
        bin_edges = np.linspace(min_val, max_val, num_bins + 1)
    
    # Criar labels customizados com nome da coluna e range de valores
    labels = []
    for i in range(len(bin_edges) - 1):
        min_val = int(bin_edges[i])
        max_val = int(bin_edges[i + 1])
        label = f'{col}_{min_val}-{max_val}'
        labels.append(label)
    
    # Usar pd.cut com os bins obtidos para criar a categorização com labels customizados
    df_eja[col_cat] = pd.cut(col_numeric, bins=bin_edges, labels=labels, include_lowest=True, duplicates='drop')
    
    print(f'  {col} -> {col_cat} (faixas: {df_eja[col_cat].nunique()})')

print(f'\nDataFrame final: {df_eja.shape[0]} registros x {df_eja.shape[1]} colunas')
print(f'Exemplo de distribuição de faixas para QT_MAT_BAS:')
print(df_eja['QT_MAT_BAS_FAIXA'].value_counts().sort_index())
```

```python
# Imprimir o novo dataframe com as colunas de faixas
print(f'DataFrame final: {df_eja.shape[0]} registros x {df_eja.shape[1]} colunas')
print(f'\nColunas do dataframe:')
print(df_eja.columns.tolist())
print(f'\nPrimeiras 5 linhas do dataframe:')
df_eja.head()
```

```python
# Converter NU_ANO_CENSO para category
df_eja['NU_ANO_CENSO'] = df_eja['NU_ANO_CENSO'].astype('category')
print(f'NU_ANO_CENSO convertido para category: {df_eja["NU_ANO_CENSO"].dtype}')
```

```python
# Treinar modelo de Association Rules usando colunas categóricas
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Selecionar apenas colunas categóricas (NU_ANO_CENSO e colunas _FAIXA)
colunas_categoricas = ['NU_ANO_CENSO'] + [col for col in df_eja.columns if col.endswith('_FAIXA')]
print(f'Colunas categóricas selecionadas: {len(colunas_categoricas)}')

# Criar dataset para association rules (one-hot encoding)
df_categorical = df_eja[colunas_categoricas].copy()

# Converter para string para garantir consistência
for col in df_categorical.columns:
    df_categorical[col] = df_categorical[col].astype(str)

# Remover linhas com NaN
df_categorical = df_categorical.dropna()
print(f'Dataset após remover NaN: {df_categorical.shape[0]} registros')

# Criar one-hot encoding manual
df_onehot = pd.get_dummies(df_categorical, prefix='', prefix_sep='')
print(f'One-hot encoding: {df_onehot.shape[0]} registros x {df_onehot.shape[1]} features')

# Aplicar algoritmo Apriori
min_support = 0.01  # Suporte mínimo de 1%
frequent_itemsets = apriori(df_onehot, min_support=min_support, use_colnames=True, max_len=3)
print(f'Itemsets frequentes encontrados: {len(frequent_itemsets)}')

# Gerar regras de associação
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=1.0)
print(f'Regras de associação geradas: {len(rules)}')

# Imprimir as 10 regras mais relevantes (por lift)
print('\n=== 10 REGRAS DE ASSOCIAÇÃO MAIS RELEVANTES (por lift) ===')
top_rules = rules.sort_values('lift', ascending=False).head(10)
for idx, row in top_rules.iterrows():
    antecedents = ', '.join([str(a) for a in row['antecedents']])
    consequents = ', '.join([str(c) for c in row['consequents']])
    print(f'\nRegra {idx}:')
    print(f'  Se {antecedents} então {consequents}')
    print(f'  Suporte: {row["support"]:.4f}')
    print(f'  Confiança: {row["confidence"]:.4f}')
    print(f'  Lift: {row["lift"]:.4f}')
```
