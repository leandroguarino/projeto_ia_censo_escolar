```python
import pandas as pd

# Caminho do arquivo CSV
caminho_arquivo = '../data/microdados_censo_da_educacao_superior_2024/dados/MICRODADOS_ED_SUP_IES_2024.CSV'

# Ler o arquivo CSV
df = pd.read_csv(caminho_arquivo, sep=';', encoding='latin-1')

# Imprimir as colunas do arquivo
print("Colunas do arquivo:")
print(df.columns.tolist())

# Imprimir informações básicas
print(f"\nNúmero de linhas: {len(df)}")
print(f"Número de colunas: {len(df.columns)}")
```

```python
import matplotlib.pyplot as plt

# Contar a quantidade de IES por estado
ies_por_uf = df['SG_UF_IES'].value_counts().sort_values(ascending=False)

# Plotar gráfico de barras
plt.figure(figsize=(12, 6))
bars = plt.bar(ies_por_uf.index, ies_por_uf.values)
plt.title('Quantidade de IES por Estado (UF)')
plt.xlabel('Estado (UF)')
plt.ylabel('Quantidade de IES')
plt.xticks(rotation=45)

# Adicionar labels em cada barra
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height)}',
             ha='center', va='bottom')

plt.tight_layout()
plt.show()

# Imprimir os dados
print("\nQuantidade de IES por estado:")
print(ies_por_uf)
```

```python
# Caminho do arquivo CSV de cursos
caminho_cursos = '../data/microdados_censo_da_educacao_superior_2024/dados/MICRODADOS_CADASTRO_CURSOS_2024.CSV'

# Ler o arquivo CSV de cursos
df_cursos = pd.read_csv(caminho_cursos, sep=';', encoding='latin-1')

# Imprimir as colunas do arquivo de cursos
print("Colunas do arquivo de cursos:")
print(df_cursos.columns.tolist())

# Verificar se CO_IES está presente
if 'CO_IES' in df_cursos.columns:
    print("\n✓ CO_IES encontrado no dataset de cursos (chave estrangeira)")
else:
    print("\n✗ CO_IES não encontrado no dataset de cursos")

# Imprimir informações básicas
print(f"\nNúmero de linhas: {len(df_cursos)}")
print(f"Número de colunas: {len(df_cursos.columns)}")
```

```python
# Merge entre os dataframes de IES e cursos usando CO_IES
df_merged = pd.merge(df_cursos, df[['CO_IES', 'SG_UF_IES']], on='CO_IES', how='left')

# Imprimir informações do merge
print(f"DataFrame mergeado: {len(df_merged)} linhas")
print(f"Colunas disponíveis no merge: {len(df_merged.columns)}")
```

```python
# Merge de dados por estado: QT_ING, QT_ING_0_17, QT_ING_18_24 (2024) e QT_MAT_MED (2023)

# Carregar dados do Censo Escolar 2023 para QT_MAT_MED
df_2023 = pd.read_csv('../data/microdados_censo_escolar_2023/dados/microdados_ed_basica_2023.csv', encoding='latin-1', sep=';', low_memory=False)

# Agrupar QT_MAT_MED por estado (2023)
mat_med_2023 = df_2023.groupby('SG_UF')['QT_MAT_MED'].sum().reset_index()
mat_med_2023.columns = ['SG_UF', 'QT_MAT_MED_2023']

# Carregar dados do Censo Superior 2024 para QT_ING
df_ies_2024 = pd.read_csv('../data/microdados_censo_da_educacao_superior_2024/dados/MICRODADOS_ED_SUP_IES_2024.CSV', sep=';', encoding='latin-1')
df_cursos_2024 = pd.read_csv('../data/microdados_censo_da_educacao_superior_2024/dados/MICRODADOS_CADASTRO_CURSOS_2024.CSV', sep=';', encoding='latin-1', low_memory=False)

# Merge IES e cursos 2024
df_merged_2024 = pd.merge(df_cursos_2024, df_ies_2024[['CO_IES', 'SG_UF_IES']], on='CO_IES', how='left')

# Agrupar QT_ING, QT_ING_0_17, QT_ING_18_24 por estado (2024)
ing_2024 = df_merged_2024.groupby('SG_UF_IES')[['QT_ING', 'QT_ING_0_17', 'QT_ING_18_24']].sum().reset_index()
ing_2024.columns = ['SG_UF', 'QT_ING_2024', 'QT_ING_0_17_2024', 'QT_ING_18_24_2024']

# Merge dos dois datasets por estado
df_merge_estado = pd.merge(mat_med_2023, ing_2024, on='SG_UF', how='outer')

# Preencher NaN com 0
df_merge_estado = df_merge_estado.fillna(0)

print("Dados mesclados por estado:")
print(df_merge_estado)

print("\nResumo estatístico:")
print(df_merge_estado.describe())
```

```python
# Calcular porcentagem de alunos do ensino médio (2023) que ingressam no ensino superior (2024)

# Calcular total de ingressantes no ensino superior (0-17 + 18-24 anos)
df_merge_estado['QT_ING_TOTAL_2024'] = df_merge_estado['QT_ING_0_17_2024'] + df_merge_estado['QT_ING_18_24_2024']

# Calcular porcentagem de transição do ensino médio para superior
df_merge_estado['PORC_TRANSICAO_MED_SUP'] = (df_merge_estado['QT_ING_TOTAL_2024'] / df_merge_estado['QT_MAT_MED_2023']) * 100

# Ordenar por porcentagem
df_merge_estado_sorted = df_merge_estado.sort_values('PORC_TRANSICAO_MED_SUP', ascending=False)

print("Porcentagem de alunos do ensino médio (2023) que ingressaram no ensino superior (2024):")
print(df_merge_estado_sorted[['SG_UF', 'QT_MAT_MED_2023', 'QT_ING_TOTAL_2024', 'PORC_TRANSICAO_MED_SUP']].to_string(index=False))

# Criar gráfico de barras
plt.figure(figsize=(14, 8))
bars = plt.bar(df_merge_estado_sorted['SG_UF'], df_merge_estado_sorted['PORC_TRANSICAO_MED_SUP'], color='#2E86AB', alpha=0.7)

plt.title('Porcentagem de Alunos do Ensino Médio (2023) que Ingressaram no Ensino Superior (2024)', fontsize=14, fontweight='bold')
plt.xlabel('Estado (UF)', fontsize=12)
plt.ylabel('Porcentagem de Transição (%)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

# Adicionar labels com o valor da porcentagem em cada barra
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.1f}%',
             ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

# Exibir estatísticas
print(f"\n{'='*70}")
print(f"ESTATÍSTICAS - TRANSIÇÃO ENSINO MÉDIO PARA SUPERIOR (2023 -> 2024)")
print(f"{'='*70}")
print(f"\nMédia: {df_merge_estado['PORC_TRANSICAO_MED_SUP'].mean():.2f}%")
print(f"Mediana: {df_merge_estado['PORC_TRANSICAO_MED_SUP'].median():.2f}%")
print(f"Desvio Padrão: {df_merge_estado['PORC_TRANSICAO_MED_SUP'].std():.2f}%")
print(f"Valor Mínimo: {df_merge_estado['PORC_TRANSICAO_MED_SUP'].min():.2f}%")
print(f"Valor Máximo: {df_merge_estado['PORC_TRANSICAO_MED_SUP'].max():.2f}%")
```

```python
# Salvar df_merge_estado em arquivo CSV
caminho_saida = '../data/df_merge_estado_2023_2024.csv'
df_merge_estado.to_csv(caminho_saida, index=False, encoding='utf-8')

print(f"DataFrame salvo com sucesso em: {caminho_saida}")
print(f"Número de linhas: {len(df_merge_estado)}")
print(f"Número de colunas: {len(df_merge_estado.columns)}")
print(f"\nColunas salvas: {df_merge_estado.columns.tolist()}")
```

```python
# Gráficos: Top 10 IES com mais matriculados em SC e PR (barras empilhadas por modalidade)

# Incluir NO_IES no merge
df_merged_com_nome = pd.merge(df_cursos, df[['CO_IES', 'SG_UF_IES', 'NO_IES']], on='CO_IES', how='left')

# Filtrar por modalidade: 1 = Presencial, 2 = EAD
df_presencial = df_merged_com_nome[df_merged_com_nome['TP_MODALIDADE_ENSINO'] == 1]
df_ead = df_merged_com_nome[df_merged_com_nome['TP_MODALIDADE_ENSINO'] == 2]

# Filtrar e agrupar para SC
sc_presencial = df_presencial[df_presencial['SG_UF_IES'] == 'SC'].groupby('NO_IES')['QT_MAT'].sum()
sc_ead = df_ead[df_ead['SG_UF_IES'] == 'SC'].groupby('NO_IES')['QT_MAT'].sum()
sc_total = sc_presencial.add(sc_ead, fill_value=0)
sc_top10 = sc_total.sort_values(ascending=False).head(10)

# Filtrar e agrupar para PR
pr_presencial = df_presencial[df_presencial['SG_UF_IES'] == 'PR'].groupby('NO_IES')['QT_MAT'].sum()
pr_ead = df_ead[df_ead['SG_UF_IES'] == 'PR'].groupby('NO_IES')['QT_MAT'].sum()
pr_total = pr_presencial.add(pr_ead, fill_value=0)
pr_top10 = pr_total.sort_values(ascending=False).head(10)

# Criar figura com 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

# Gráfico SC - Barras empilhadas
sc_presencial_vals = sc_top10.index.map(lambda x: sc_presencial.get(x, 0))
sc_ead_vals = sc_top10.index.map(lambda x: sc_ead.get(x, 0))

bars1 = ax1.barh(sc_top10.index, sc_presencial_vals, label='Presencial', color='#4A90E2', alpha=0.7)
bars1_ead = ax1.barh(sc_top10.index, sc_ead_vals, left=sc_presencial_vals, label='EAD', color='#FF6B6B', alpha=0.7)

ax1.set_title('Top 10 IES - Santa Catarina (SC)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Total de Matriculados', fontsize=12)
ax1.invert_yaxis()
ax1.legend(loc='lower right')
ax1.grid(axis='x', alpha=0.3)

# Adicionar labels com total
for i, (ies, total) in enumerate(sc_top10.items()):
    ax1.text(total + 5000, i, f'{int(total):,}', 
             ha='left', va='center', fontsize=10, fontweight='bold')

# Gráfico PR - Barras empilhadas
pr_presencial_vals = pr_top10.index.map(lambda x: pr_presencial.get(x, 0))
pr_ead_vals = pr_top10.index.map(lambda x: pr_ead.get(x, 0))

bars2 = ax2.barh(pr_top10.index, pr_presencial_vals, label='Presencial', color='#4CAF50', alpha=0.7)
bars2_ead = ax2.barh(pr_top10.index, pr_ead_vals, left=pr_presencial_vals, label='EAD', color='#FFA500', alpha=0.7)

ax2.set_title('Top 10 IES - Paraná (PR)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Total de Matriculados', fontsize=12)
ax2.invert_yaxis()
ax2.legend(loc='lower right')
ax2.grid(axis='x', alpha=0.3)

# Adicionar labels com total
for i, (ies, total) in enumerate(pr_top10.items()):
    ax2.text(total + 5000, i, f'{int(total):,}', 
             ha='left', va='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print("\nTop 10 IES - Santa Catarina (SC):")
print(f"{'IES':<50} {'Presencial':<12} {'EAD':<12} {'Total':<12}")
print("-" * 88)
for ies in sc_top10.index:
    pres = sc_presencial.get(ies, 0)
    ead = sc_ead.get(ies, 0)
    total = pres + ead
    print(f"{ies:<48} {int(pres):<12,} {int(ead):<12,} {int(total):<12,}")

print("\nTop 10 IES - Paraná (PR):")
print(f"{'IES':<50} {'Presencial':<12} {'EAD':<12} {'Total':<12}")
print("-" * 88)
for ies in pr_top10.index:
    pres = pr_presencial.get(ies, 0)
    ead = pr_ead.get(ies, 0)
    total = pres + ead
    print(f"{ies:<48} {int(pres):<12,} {int(ead):<12,} {int(total):<12,}")
```
